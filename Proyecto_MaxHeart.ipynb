```python
# CONTENEDORES DIGITALES, INDEPENDIENTES Y MUTABLES PARA EL CONTROL DE MAXHEART

# Creamos cuatro diccionarios vacíos y limpios, usando llaves estructuradas.
# Cada uno almacenará de forma aislada los registros según su tipo de solicitud.

diccionario_peticiones = {}
diccionario_quejas = {}
diccionario_reclamos = {}
diccionario_sugerencias = {}

# ID de registro: número entero auto-incremental único e inalterable (RF2)
id_consecutivo_radicado = 1

```

```python
# DATOS DE CIUDADANO SOLICITANTE.

# Usamos la función input de forma directa para interactuar con el administrador

print("---INGRESO DE NUEVA PQRS EN EL SISTEMA MAXHEART---")

# Solicitamos el nombre obligatorio aplicando métodos de limpieza de cadenas.
nombre_solicitante = input("Nombre completo del ciudadano: ").strip().title()

# Registro estricto del tipo de documento obligatorio
tipo_documento_id = input("Tipo de documento (CC, TI, CE, PP, NIT): ").strip().upper()
numero_documento_id = input("Número de documento de identidad: ").strip()

# Datos de contacto, teléfonos y correos electrónicos obligatorios
tipo_telefono_contacto = input("Tipo de teléfono (Celular, Fijo, Corporativo): ").strip().capitalize()
numero_telefono_contacto = input("Número telefónico de contacto (10 dígitos): ").strip()


# RF1: Solicitud y validación institucional mandatoria del correo electrónico.
correo_electronico_usuario = input("Correo electrónico institucional (@udea.edu.co): ").strip().lower()

# Operaciones de cadena para verificar un único arroba y la extensión del dominio oficial.
cantidad_arrobas_correo = correo_electronico_usuario.count("@")
tiene_dominio_udea = correo_electronico_usuario.endswith("@udea.edu.co")

#Estructura condicional simple de toma de decisiones para restringir el ingreso
if cantidad_arrobas_correo == 1 and tiene_dominio_udea == True:
   print("Validación exitosa: Correo electronico institucional certificado")

   # Requerimiento opcional: dirección de correspondencia domiciliaria
   direccion_residencia = input("Dirección de correspondencia (Opcional): ").strip()

   # Validación simple de campo opcional: si se deja vacío, se asigna "N/A"
   if direccion_residencia == "":
    direccion_residencia = "N/A"

else:
     print("ERROR FATAL: El correo ingresado no pertenece al dominio institucional válido.")
     # Forzamos la anulación de las variables para evitar un NameError posterior
     correo_electronico_usuario = "CORREO_INVÁLIDO"
     direccion_residencia = "N/A"
```

```python
# CLASIFICACIÓN DE LA PQRS Y GESTIÓN AUTOMÁTICA DE TIEMPOS EXTREMOS - MAXHEART.

# Realizaremos la clasificación de la especie y el cálculo de la fecha límite (RF4)

# 1. Clasificación del trámite de forma estandarizada.
tipo_solicitud_elegida = input("Tipo de PQRS (Petición, Queja, Reclamo, Sugerencia): ").strip().capitalize()
canal_recepcion_pqrs = input("Canal de recepción (Página web, Correo, Presencial): ").strip().capitalize()

# RF3: TIPIFICACIÓN OBLIGATORIA DE ESPECIE RESTRICTIVA.
tipo_mascota_paciente = input("Tipo de mascota afectada (Perro/Gato): ").strip().capitalize()
campus_universitario_relacionado = input("Campus UdeA vinculado al caso: ").strip().title()

asunto_titulo_breve = input("Asunto o título resumido de la solicitud: ").strip().capitalize()
descripcion_detallada_caso = input("Descripción detallada del requerimiento: ").strip()

# 2. Operaciones de tiempo con la librería DateTime.
# Se importa la herramienta para poder realizar cálculos aritméticos de fechas.
from datetime import datetime, timedelta

# RF4: Se captura el instante de registro y se suman con exactitud 30 días calendario
fecha_radicacion_actual = datetime.now()
fecha_limite_respuesta = fecha_radicacion_actual + timedelta(days=30)
estado_inicial_solicitud = "Registrada"

# Se convierten los datos de tiempo a cadenas estáticas limpias para la visualización.
fecha_radicacion_texto = fecha_radicacion_actual.strftime("%Y-%m-%d")
fecha_limite_texto = fecha_limite_respuesta.strftime("%Y-%m-%d")
```

```python
# EMPAQUETAMIENTO EN MEMORIA E IMPRESIÓN DEL COMPROBANTE DE RADICACIÓN.

# Se agrupan todas las variables validadas en una estructura de lista ordenada

datos_completos_radicado = [nombre_solicitante, tipo_documento_id, numero_documento_id, numero_telefono_contacto, correo_electronico_usuario, direccion_residencia, tipo_solicitud_elegida, canal_recepcion_pqrs, tipo_mascota_paciente, campus_universitario_relacionado, asunto_titulo_breve, descripcion_detallada_caso, fecha_radicacion_texto, fecha_limite_texto, estado_inicial_solicitud]

# Se guarda el resgistro en el contenedor correspondiente, usando el ID único (RF2)
if tipo_solicitud_elegida == "Petición":
   diccionario_peticiones[id_consecutivo_radicado] = datos_completos_radicado
elif tipo_solicitud_elegida == "Queja":
   diccionario_quejas[id_consecutivo_radicado] = datos_completos_radicado
elif tipo_solicitud_elegida == "Reclamo":
   diccionario_reclamos[id_consecutivo_radicado] = datos_completos_radicado
elif tipo_solicitud_elegida == "Sugerencia":
   diccionario_sugerencias[id_consecutivo_radicado] = datos_completos_radicado

# GENERACIÓN DE SALIDA SIMÉTRICA Y CONSISTENTE (Ancho fijo de 100 caracteres)
print("\n" + "=" * 100)
print("COMPROBANTE OFICIAL DE RADICACIÓN - SISTEMA DIGITAL MAXHEART UdeA".center(100, " "))
print("=" * 100)
print(f"Radicado ID: {id_consecutivo_radicado:<15} | Fecha Registro: {fecha_radicacion_texto:<15} | Estado: {estado_inicial_solicitud:<15} ")
print("-" * 100)

# DATOS EXCLUSIVOS DEL CIUDADANO SOLICITANTE
print(" Datos del ciudadano solicitante ".center(100,"-"))
print(f"Solicitante: {nombre_solicitante:<35} | Identificación: {tipo_documento_id} - {numero_documento_id:<35} ")
print(f"Contacto Tel: {numero_telefono_contacto:<32} | Correo Institucional: {correo_electronico_usuario:<35} ")
print(f"Residencia: {direccion_residencia:<83} ")

# DATOS EXCLUSIVOS DE LA MASCOTA Y EL CASO RELACIONADO
print(" Detalles de la mascota y el caso ".center(100, "-"))
print(f"Tipo de mascota: {tipo_mascota_paciente:<33} | Tipo de solicitud: {tipo_solicitud_elegida:<35} ")
print(f"Sede/Campus: {campus_universitario_relacionado:<33} | Canal: {canal_recepcion_pqrs:<35} ")
print(f"Asunto PQRS: {asunto_titulo_breve:<33} | Límite Máximo: {fecha_limite_texto:<35} ")
print("=" * 100)



# El consecutivo único se incrementa de forma automática para el próximo ciudadano (RF2)
id_consecutivo_radicado += 1
```

```python
# EXTRACCIÓN ANALÍTICA, COMPACTA Y EQUITATIVA MEDIANTE COMPRENSIÓN DE LISTAS.

# Se unifican todos los regstros en una sola estructura lineal y limpia, para luego segmentar de forma compacta por la tipificación de especies (RF3)

# Unificación horizontal de todos los valores almacenados en los 4 contenedores.
todas_las_pqrs_del_sistema = (list(diccionario_peticiones.values()) + list(diccionario_quejas.values()) + list(diccionario_reclamos.values()) + list(diccionario_sugerencias.values()))

# Revisión de casilla: Se filtra accediendo a la posición [8] de la lista de datos.
# Se extrae el nombre del ciudadano (Posición) para generar el reporte real.
global_solicitudes_perros = [pqrs[0] for pqrs in todas_las_pqrs_del_sistema if pqrs[8] == "Perro"]
global_solicitudes_gatos = [pqrs[0] for pqrs in todas_las_pqrs_del_sistema if pqrs[8] == "Gato"]

# GENERACIÓN DEL REPORTE ESTADÍSTICO UNIFICADO (Ancho de 100 caracteres)
print("\n" + "=" * 100)
print(" REPORTE GLOBAL UNIFCADO DE CONTROL DE ESPECIES - MAXHEART UdeA".center(100, " "))
print("=" * 100)
print(f" Total de ciudadanos con solicitudes asociadas a Caninos (Perros): {global_solicitudes_perros} ")
print(f" Total de ciudadanos con solicitudes asociadas a Felinos (Gatos): {global_solicitudes_gatos} ")
print("=" * 100)
```
